```python
#merge with guide metadata
guide_act = pd.read_excel('Supplementary_Table1.xlsx', sheet_name='Table S4 Guide Activity')
```

```python
keep = ['guide_identity','guide_label', 'gene_id',
       'ctrl_pctl_fibro', 'target_pctl_fibro',
       'pctl_change_fibro', 'adj_p_fibro', 'ctrl_coef_fibro', 'ctrl_abs_fibro',
       'target_coef_fibro', 'target_abs_fibro', 'count_profile_fibro',
       'ctrl_expressed_fibro', 'is_activated_fibro', 'guide_rank_fibro',
       'top_guide_fibro', 'activation_expression_fibro',
       'cluster_description_fibro', 'masked_cluster_description_fibro',
       'bulk_expr_fibro', 'expressed_fibro', 'p_expressed_fibro',
       'cell_count_fibro', 'ctrl_pctl_rpe1', 'target_pctl_rpe1',
       'pctl_change_rpe1', 'adj_p_rpe1', 'ctrl_coef_rpe1', 'ctrl_abs_rpe1',
       'target_coef_rpe1', 'target_abs_rpe1', 'count_profile_rpe1',
       'ctrl_expressed_rpe1', 'is_activated_rpe1', 'guide_rank_rpe1',
       'top_guide_rpe1', 'activation_expression_rpe1',
       'cluster_description_rpe1', 'masked_cluster_description_rpe1',
       'bulk_expr_rpe1', 'expressed_rpe1', 'p_expressed_rpe1',
       'cell_count_rpe1', 'bad_seed', 'seed_matches', 'good_guide_fibro',
       'good_guide_rpe1', 'best_guide_fibro', 'best_guide_rpe1',
       'seed_driven_fibro', 'seed_driven_rpe1', 'active_fibro',
       'expanded_active_fibro', 'de_genes_fibro', 'active_rpe1',
       'expanded_active_rpe1', 'de_genes_rpe1', 'final_tf_library']

adata.obs = adata.obs.merge(guide_act[keep], left_index = True, right_on = 'guide_identity', how = 'left').set_index('guide_identity')
```

```python
hits_fibro = adata.obs[(adata.obs.seed_driven_fibro == False)&(adata.obs.bad_seed == False)&((adata.obs.is_activated_fibro >0)|(adata.obs.active_fibro)|(adata.obs.expanded_active_fibro))]
```
